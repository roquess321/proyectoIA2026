# Datasets Públicos

```
from sklearn.datasets.openml import fetch_openml
import tempfile
test_data = tempfile.mkdtemp()

# www.openml.org

iris = fetch_openml('iris', 1, data_home = test_data)
```

```
print(iris.target.shape)
print(iris.data.shape)
print(iris.feature_names)
```

```
electricity = fetch_openml('electricity',1, data_home = test_data)
```

```
print(electricity.target.shape)
print(electricity.data.shape)
print(electricity.feature_names)
```
