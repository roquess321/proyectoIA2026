# Clases

```python
class Alumno:

    def saludar(self):
        """Imprime un saludo en pantalla."""
        print("¡Hola, mundo!")
```

```python
alumno = Alumno()
```

```python
alumno.saludar()
```

```python
class Alumno:

    def __init__(self):
        self._nombre = "Pablo"

    def saludar(self):
        """Imprime un saludo en pantalla."""
        print("¡Hola, {x}!".format(x = self._nombre))
```

```python
alumno = Alumno()
```

```python
alumno.saludar()
```

```python
class Persona:

    def __init__(self, nombre='José', edad = 28):
        self._nombre = nombre
        self._edad = edad
```

```python
class Profesor(Persona):

    def __init__(self, nombre, edad, asignatura):
        super().__init__(nombre, edad)
        self._asignatura = asignatura

    def presentar(self):
      print(self._asignatura.split())
      print(self._edad**2)
      print('Buenas, mi nombre es {nom}, tengo {ed} años y enseñó la asignatura {asig}'.format(nom=self._nombre, ed=self._edad, asig=self._asignatura))
```

```python
profesor = Profesor('Enrique', 28, 'Inteligencia Artificial')
```

```python
profesor.presentar()
```
